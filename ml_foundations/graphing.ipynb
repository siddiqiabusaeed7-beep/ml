```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
iris = pd.read_csv("iris.csv")
```

```python
#sampe
iris.sample(5)
```

replacing. : the parameters is passed in the form of a dictionary
to change the value of a column

```python
"""replacing. : the parameters is passed in the form of a dictionary"""

iris['Species']= iris['Species'].replace({'Iris-setosa' : 0, 'Iris-virginica':1, 'Iris-versicolor':2})
iris.sample(5)
```

```python
"""c : used to assign random colours to different graph classes
    cmap is like the themes of teh colour
    alpha: to control the opcity of the graph"""

plt.scatter(iris["SepalLengthCm"], iris["PetalLengthCm"], c=iris["Species"], cmap= 'winter', alpha =0.6)
plt.xlabel("SepalLengthCm")
plt.ylabel("PetalLengthCm")
plt.colorbar()
```

PLOT SIZE

```python
plt.figure(figsize =(15,7))
plt.scatter(iris["SepalLengthCm"], iris["PetalLengthCm"], c=iris["Species"], cmap= 'winter', alpha =0.6)
plt.xlabel("SepalLengthCm")
plt.ylabel("PetalLengthCm")
plt.colorbar()
```

ANNOTATIONS

placing a text above a point in the graph

```python
batsman = pd.read_csv("batter.csv")
batsman.head()
```

```python
sample_df = batsman.head(100).sample(25, random_state= 5)
"""taking the top 100 batsman and then selecting random 25 batsman"""
```

```python
sample_df
```

```python
plt.scatter(sample_df["avg"], sample_df["strike_rate"])
```

```python
plt.scatter(sample_df["avg"], sample_df["strike_rate"], s= sample_df["runs"])
for i in range(sample_df.shape[0]):
    plt.text(sample_df["avg"].values[i], sample_df["strike_rate"].values[i], sample_df["batter"].values[i])

plt.figure(figsize=(40,30))
```

#HORIZONTAL AND VERTICAL LINES

```python
plt.scatter(sample_df["avg"], sample_df["strike_rate"], s= sample_df["runs"])
plt.axhline(130 ,color="red")
plt.axhline(140, color="green")
plt.axvline(30, color='red')

for i in range(sample_df.shape[0]):
    plt.text(sample_df["avg"].values[i], sample_df["strike_rate"].values[i], sample_df["batter"].values[i])

plt.figure(figsize=(40,30))
```

#SUBPLOTS

```python
fig, ax = plt.subplots()

ax.scatter(batsman["runs"], batsman["avg"])
```

```python
fig , ax = plt.subplots(nrows=2, ncols= 1, sharex =True)

ax[0].scatter(batsman["runs"], batsman["avg"])
ax[1].scatter(batsman["runs"], batsman["strike_rate"])

ax[0].set_title("runs vs average")
ax[0].set_ylabel("avg")

ax[1].set_title("Runs vs Strike Rate")
ax[1].set_ylabel("strike rate")
plt.show()
```

```python
fig, ax = plt.subplots(nrows=2, ncols = 2)

ax[0,0].scatter(batsman["runs"], batsman["avg"])
ax[1,0].scatter(batsman["runs"], batsman["strike_rate"])
ax[0,1].hist(batsman["avg"])
ax[1,1].hist(batsman["runs"])

ax[0,0].set_title("runs vs average")
ax[0,0].set_ylabel("avg")

ax[1,0].set_title("Runs vs Strike Rate")
ax[1,0].set_ylabel("strike rate")

ax[0,1].set_title("Average")
ax[1,1].set_title("Runs")
```

another method to do so

```python
"""creating a figure object"""

fig = plt.figure()
ax1 =  fig.add_subplot(1,2,1) #(row, column , position)
ax1.scatter(batsman["runs"], batsman["avg"])

ax2 = fig.add_subplot(1,2,2)
ax2.scatter(batsman["runs"], batsman["strike_rate"])
```

3D SURFACE PLOTS

```python
x = np.linspace(-10, 10, 100)
y= np.linspace(-10, 10,100)
```

```python
x
```

```python
xx, yy = np.meshgrid(x, y)
```

```python
z = xx**2 + yy**2
```

```python
z
```

```python
fig = plt.figure(figsize= (12,8))

ax = plt.subplot(projection="3d")
p = ax.plot_surface(xx,yy,z, cmap = "viridis")
plt.colorbar(p)
```

```python
fig = plt.figure(figsize= (12,8))
fig = plt.subplot(projection="3d")
p = fig.plot_surface(xx,yy,z, cmap = "viridis")
plt.colorbar(p)
```

#CONTOUR PLOT

```python
fig = plt.figure()

ax= plt.subplot()
p = ax.contour(xx,yy,z,cmap="viridis")
fig.colorbar(p)
```

```python
fig = plt.figure()

ax= plt.subplot()
p = ax.contourf(xx,yy,z,cmap="viridis")
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig=plt.figure()

ax = plt.subplot(projection="3d")
p = ax.plot_surface(xx, yy, z, cmap = 'viridis')
fig.colorbar(p)
                 
```
